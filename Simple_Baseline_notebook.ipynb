```python
"""goal for now write a preprocessing function that creates a 
new product category that merges the product category to thier 
proper place (removes duplicates)then use that product_category 
to find the mean total_sales then map that mean total_sales then use it for the baseline"""
```

```python
# written using AI Code review later for 
def clean_product_category(df, source_col, target_col):
    """
    Normalizes case/whitespace duplicates in a categorical column
    (e.g. 'seafood', 'SEAFOOD', 'SeaFood' -> 'seafood') and writes
    the result into a new column.
    """
    df[target_col] = (
        df[source_col]
        .astype(str)
        .str.strip()
        .str.lower()
    )
    return df
```

```python
train = clean_product_category(train, 'product_category', 'cleaned_product_category')

# sanity check: how many unique categories now vs before
print(train['product_category'].nunique(), '->', train['cleaned_product_category'].nunique())
print(train['cleaned_product_category'].value_counts())
```

### mean_baseline function and score 

```python
def category_mean_baseline(train, test, cat_col='product_category', 
                             cleaned_col='cleaned_product_category', 
                             target_col='total_sales'):
    """
    Baseline model: predicts total_sales as the mean total_sales
    of the (cleaned) product category, learned from train and
    applied to test.

    Returns:
        test  -> test df with cleaned_col and a new 'predicted_<target_col>' column
        mean_store_sales -> the train-derived mapping (Series), sorted desc
    """
    # 1. clean train categories
    train = clean_product_category(train, cat_col, cleaned_col)

    # 2. learn the mapping from train only
    mean_store_sales = (
        train.groupby(cleaned_col)[target_col]
        .mean()
        .sort_values(ascending=False).round(2)
    )

    # 3. clean test categories the same way
    test = clean_product_category(test, cat_col, cleaned_col)

    # 4. map train's mean onto test's cleaned category
    #    fallback: overall train mean for any category unseen in train
    overall_mean = train[target_col].mean().round(2)
    test[f'predicted_mean_{target_col}'] = (
        test[cleaned_col].map(mean_store_sales).fillna(overall_mean).round(2)
    )

    return test, mean_store_sales, overall_mean
```

```python
# Example usage
test, mean_store_sales, overall_mean = category_mean_baseline(train, test)
print(mean_store_sales)
print(test[['product_category', 'cleaned_product_category', f'predicted_mean_total_sales']].head())
```

### median_baseline function and checking 

```python
def category_median_baseline(train, test, cat_col='product_category', 
                             cleaned_col='cleaned_product_category', 
                             target_col='total_sales'):
    """
    Baseline model: predicts total_sales as the median total_sales
    of the (cleaned) product category, learned from train and
    applied to test.

    Returns:
        test  -> test df with cleaned_col and a new 'predicted_<target_col>' column
        median_store_sales -> the train-derived mapping (Series), sorted desc
    """
    # 1. clean train categories
    train = clean_product_category(train, cat_col, cleaned_col)

    # 2. learn the mapping from train only
    median_store_sales = (
        train.groupby(cleaned_col)[target_col]
        .median()
        .sort_values(ascending=False)
    )

    # 3. clean test categories the same way
    test = clean_product_category(test, cat_col, cleaned_col)

    # 4. map train's mean onto test's cleaned category
    #    fallback: overall train median for any category unseen in train
    overall_median = train[target_col].median()
    test[f'predicted_median_{target_col}'] = (
        test[cleaned_col].map(median_store_sales).fillna(overall_median)
    )

    return test, median_store_sales, overall_median
```

```python
# Example usage
test, median_store_sales, overall_median = category_median_baseline(train, test)
print(median_store_sales)
print(test[['product_category', 'cleaned_product_category', f'predicted_median_total_sales']].head())
```

```python
sample_median_df = sample

sample_median_df["total_sales"] = test["predicted_median_total_sales"]
```

```python
sample_median_df.to_csv("median_baseline.csv", index=False)
```
